# Домашнее задание №8: VQ-GAN
## Описание задачи
Реализовать VQ-GAN (Vector Quantized Generative Adversarial Network) и применить её к задаче реконструкции выбранного датасета в сравнении с обычным VQ-VAE&

Можно использовать облегченные варианты архитектуры Encoder/Decoder.  
Работаем с низким разрешением (64×64 или 128×128).  

Обязательные компоненты архитектуры:
- VQ-VAE c дискретным латентным пространство через codebook
- PatchGAN discriminator — оценка реалистичности локальных патчей

## Датасет - любой!
Я советую отойти от привычных ImageNet/COCO/Celeba и выбрать что-то креативное (Например, персонажи PixelArt или Anime Faces). Датасеты можно искать, например, на Kaggle или HuggingFace, но ;елательно, чтобы там было от 5к изображений.

## Желательные составляющие решения и система оценки
### Часть 1: Baseline VQ-VAE (2 балла)
- VQ-VAE реализован правильно (архитектурно и лосс) - макс 2 балла
- VQ-VAE сошелся и делает визуально схожую реконструкцию (норм, если она чуть мутная/мыльная) - 1 балл

### Часть 2: VQ-GAN (4 балла)
- VQ-GAN реализован правильно (архитектурно и лосс) - макс 2 балла
- VQ-GAN сошелся и делает визуально близкую реконструкцию - макс 2 балла, если практически неотличимо от GT и отличимо от VQ-VAE

### Часть 3: Сравнительный анализи подход к решению (макс 3 балла)
- Гиперпараметры подобраны с умом, есть трекинг обучения, описан и обоснован выбор тех или иных решений - макс 2 балла
- Есть визуализация процесса обучения - 1 балл
- Есть визуализации самой реконструкции и кодбуков (посмотреть в семинаре 1) и проведено сравнение - 1 балл
